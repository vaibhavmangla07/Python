```python
import nltk
import ssl

# Fix SSL certificate issue on macOS
try:
    _create_unverified_https_context = ssl._create_unverified_context
except AttributeError:
    pass
else:
    ssl._create_default_https_context = _create_unverified_https_context

nltk.download('punkt_tab')
```

```python
corpus = """Hello, how are you doing today? I hope you're having a great day! Let's meet up later and discuss the project. Don't forget to bring the documents we need for the meeting. See you soon!"""
corpus
```

## Tokenization
Tokenization is the process of breaking down a text into smaller units called tokens. 

These tokens can be words, phrases, or even characters. 

Tokenization is a crucial step in natural language processing (NLP) as it allows us to analyze and manipulate text data effectively.

```python
# Tokenization
# Sentences --> Paragraphs
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(corpus)
sentences
```

```python
type(sentences)
```

```python
# Paragraphs --> Words
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
words
```

```python
# Sentences --> Words
for sentence in sentences:
    print(word_tokenize(sentence))
```

```python
# wordpunct_tokenize() --> It is used to split the text into words and punctuation. It is a more advanced version of word_tokenize() that also separates punctuation from words.
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokens = wordpunct_tokenize(corpus)
wordpunct_tokens
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
treebank_tokens = tokenizer.tokenize(corpus)
treebank_tokens
```
