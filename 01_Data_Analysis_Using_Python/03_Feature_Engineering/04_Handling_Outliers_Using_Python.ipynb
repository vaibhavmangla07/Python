## 5 Numbers Summary and Box Plots

```python
# Minimum, Maximum, Mean, Q1, Q3, IQR, Lower Whisker, Upper Whisker, Outliers, Non-Outliers, Z-Score, Standard Deviation, Variance, Percentile, Quartile, Median
```

```python
import numpy as np
```

```python
lst_marks = [45, 67, 89, 23, 78, 90, 56, 49, 100, 34, 23, 67, 88, 92, 54, 76, 81, 95, 38, 72]
print("Minimum:", np.min(lst_marks))
print("Maximum:", np.max(lst_marks))
print("Mean:", np.mean(lst_marks))
print("Median:", np.median(lst_marks))
print("Standard Deviation:", np.std(lst_marks))
print("Variance:", np.var(lst_marks))    
```

```python
print("Q1 (25th Percentile):", np.percentile(lst_marks, 25))
print("Q2 (50th Percentile):", np.percentile(lst_marks, 50))
print("Q3 (75th Percentile):", np.percentile(lst_marks, 75))   
print("Q4 (100th Percentile):", np.percentile(lst_marks, 100))
```

```python
iqr = np.percentile(lst_marks, 75) - np.percentile(lst_marks, 25)
print("IQR (Interquartile Range):", np.percentile(lst_marks, 75) - np.percentile(lst_marks, 25))

lower_whisker = np.percentile(lst_marks, 25) - 1.5 * iqr
print("Lower Whisker:", lower_whisker)

upper_whisker = np.percentile(lst_marks, 75) + 1.5 * iqr
print("Upper Whisker:", upper_whisker)
```

```python
outliers = [x for x in lst_marks if x < lower_whisker or x > upper_whisker]
non_outliers = [x for x in lst_marks if x >= lower_whisker and x <= upper_whisker]
print("Outliers:", outliers)
print("Non-Outliers:", non_outliers)
```

```python
z_scores = [(x - np.mean(lst_marks)) / np.std(lst_marks) for x in lst_marks]
print("Z-Scores:", z_scores)
```
