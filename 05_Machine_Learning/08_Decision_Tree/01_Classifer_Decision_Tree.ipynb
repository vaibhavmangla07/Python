# Decision Tree Classifier

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
data = load_iris()
data
```

```python
print(data['DESCR'])
```

```python
print(data['target'])
```

```python
print(data['target_names'])
```

```python
df = pd.DataFrame(data['data'], columns=data['feature_names'])
df['target'] = data['target']
df.head()
```

```python
# independent and dependent features
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
# Visualization of Decision Tree
from sklearn import tree
plt.figure(figsize=(15, 10))
tree.plot_tree(model, filled=True, feature_names=data['feature_names'], class_names=data['target_names'])
plt.show()
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

# Decision Tree Classifier Prepruning and Hyper-Parameter Tuning

```python
param = {
    'criterion': ['gini', 'entropy', 'log_loss'],
    'splitter': ['best', 'random'],
    'max_depth': [1, 2, 3, 4, 5],
    'max_features': ['auto', 'sqrt', 'log2'],
}
param
```

```python
import warnings
warnings.filterwarnings('ignore')
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=model, param_grid=param, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)
```
