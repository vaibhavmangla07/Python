## Isolation Forest Anomaly Detection

```python
import pandas as pd
df = pd.read_csv('healthcare.csv')
df.head()
```

```python
X = df.iloc[:, 0]
y = df.iloc[:, -1]
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y)
```

```python
from sklearn.ensemble import IsolationForest
clf = IsolationForest(contamination=0.2)
clf.fit(df)
```

```python
pred = clf.predict(df)
pred
```

```python
import numpy as np
index = np.where(pred < 0)
index
```

```python
val = df.values
```

```python
index = np.where(pred < 0)
plt.scatter(X, y)
plt.scatter(val[index, 0], val[index, 1], edgecolors="r")
```
