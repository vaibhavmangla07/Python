```python
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
X, y = make_moons(n_samples=250, noise=0.05)
X[:5], y[:5]
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
plt.title('Two Moons Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
DBSCAN_model = DBSCAN(eps=0.3)
DBSCAN_model.fit(X_scaled)
```

```python
DBSCAN_model.labels_
```

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=DBSCAN_model.labels_, cmap='viridis')
plt.title('DBSCAN Clustering of Two Moons Dataset')
plt.xlabel('Feature 1 (scaled)')
plt.ylabel('Feature 2 (scaled)')
plt.show()
```

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=y, cmap='viridis')
plt.title('True Labels of Two Moons Dataset')
plt.xlabel('Feature 1 (scaled)')
plt.ylabel('Feature 2 (scaled)')
plt.show()
```
