# Map Function
map() is a built-in function used to apply a function to each element of an iterable (list, tuple, etc.).

```python
# Simple Map function
nums = [1, 2, 3, 4]
result = map(lambda x: x * 2, nums)
print(list(result))
```

```python
# map() with normal function
def square(x):
    return x * x

nums = [2, 3, 4]
list(map(square, nums))
```

```python
# map() with multiple iterables
a = [1, 2, 3]
b = [4, 5, 6]

list(map(lambda x, y: x + y, a, b))
```

```python
# map() with conversion of string list to integer list
str_nums = ['1', '2', '3', '4']
list(map(int, str_nums))
```

```python
# map() with user input
nums = list(map(int, input("Enter numbers: ").split()))
print(nums)
```

```python
# map() vs loop

# Using loop
nums = [1, 2, 3]
result = []

for i in nums:
    result.append(i * 2)
print(result)

# Using map()
nums = [1, 2, 3]
list(map(lambda x: x * 2, nums))

```

```python
# map() with string operation
words = ['hello', 'world', 'python']
list(map(lambda x: x.upper(), words))
```

```python
# map() with lambda 
a = [1, 2, 3, 4]
res = list(map(lambda x: x ** 2, a))
print(res)
```

```python
# map with list comprehension : Used to create a new list by applying an expression to each item in an iterable.
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    # print(i)
    print(i())
```

map() vs List Comprehension

| Feature      | `map()`         | List Comprehension |
| ------------ | --------------- | ------------------ |
| Readability  | Medium          | High               |
| Lambda usage | Common          | Optional           |
| Performance  | Slightly faster | Slightly slower    |
| Pythonic     | Less            | More               |
