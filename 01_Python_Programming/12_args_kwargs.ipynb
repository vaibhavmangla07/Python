# *args and **kwargs in Python

They allow functions to accept a variable number of arguments.

### *args (Non-Keyword Arguments)
✅ What is *args?

Allows function to accept multiple positional arguments

Stored as a tuple

* tells Python to pack values

```python
# Example 1 – Basic
def add(*args):
    print(args)

add(1, 2, 3)
```

```python
# Example 2 – Sum of Numbers
def total(*args):
    return sum(args)

print(total(1, 2, 3, 4))
```

```python
# Example 3 – Loop Through args
def show(*args):
    for i in args:
        print(i)

show("Python", "AI", "ML")
```

### **kwargs (Keyword Arguments)
✅ What is **kwargs?

Accepts multiple keyword arguments

Stored as a dictionary

** packs key-value pairs

```python
# Example 1 – Basic
def details(**kwargs):
    print(kwargs)

details(name="Aman", age=21)
```

```python
# Example 2 – Access Values
def show(**kwargs):
    for k, v in kwargs.items():
        print(k, v)

show(name="Ravi", city="Delhi")
```

```python
# Using Both *args and **kwargs
def demo(a, *args, **kwargs):
    print(a)
    print(args)
    print(kwargs)

demo(10, 20, 30, name="Aman")
```

```python
# Packing vs Unpacking

# Packing
def func(*args):
    print(args)

# Unpacking
nums = [1,2,3]
print(*nums)
```
