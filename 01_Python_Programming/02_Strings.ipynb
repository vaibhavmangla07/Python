# Strings

```python
str = "Hello, World!"
str
```

```python
# The len() method use to determine the length of the string.
len(str)
```

```python
# The upper() method converts a string to upper case.
str.upper()
```

```python
# The lower() method converts a string to lower case.
str.lower()
```

```python
# The strip() method removes any whitespace from the beginning or the end.
str.strip("!")  # removes exclamation marks from both ends
```

```python
# The rstrip() method removes any whitespace from the end, right side of the string.
str.rstrip("!")  # removes exclamation marks from the right end
```

```python
# The replace() method replaces a string with another string.
str.replace("World", "Python")
```

```python
# The split() method splits the give string at the specified instance and returns the separated strings as list items.
str.split(",")  # splits at the comma
```

```python
# The startswith() method returns true if the string starts with the specified value, otherwise false.
str.startswith("Hello")
```

```python
# The capitalize() method turns only the first character of the string to uppercase and the rest other characters of the string are turned to lowercase. The string has no effect if the first character is already uppercase.
str.capitalize()
```

```python
# the center() method centers the string, using a specified character (space is default) as the fill character.
str.center(50, "-")
```

```python
# the count() method returns the number of times a specified value occurs in a string.
str.count("o")
```

```python
# the endwith() method returns true if the string ends with the specified value, otherwise false.
str.endswith("World!")
```

```python
# the find() method finds the first occurrence of the specified value.
str.find("Hello")  # returns the index of first occurrence of "o"
```

```python
# the index() method finds the first occurrence of the specified value.
str.index("o")  # returns the index of first occurrence of "o"
```

```python
# the isalnum() method returns True if all characters in the string are alphanumeric (a-z, A-Z, 0-9). Example:
str.isalnum()
```

```python
# the isalpha() method returns True if all characters in the string are alphanumeric (letters and numbers), otherwise False.
str.isalpha()  # returns False because of space and punctuation
```

```python
# islower() method returns True if all characters in the string are lower case, otherwise False.
str.islower()  # returns False
```

```python
# isupper() method returns True if all characters in the string are upper case, otherwise False.
str.isupper()  # returns False 
```

```python
# isprintable() method returns True if all characters in the string are printable, otherwise False.
str.isprintable()  # returns True
```

```python
# isspace() method returns True if all characters in the string are whitespaces, otherwise False.
str.isspace()  # returns False
```

```python
# istitle() method returns True if the string follows the rules of a title, otherwise False.
str.istitle()  # returns True
```

```python
# startwith() method returns true if the string starts with the specified value, otherwise false.
str.startswith("Hello")
```

```python
# swapcase() method returns a string where all the upper case letters are converted to lower case letters and vice versa.
str.swapcase()
```

```python
# title() method converts the first character of each word to upper case.
str.title()
```
