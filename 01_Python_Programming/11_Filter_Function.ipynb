# filter() Function in Python 
✅ What is filter()?

filter() is a built-in Python function used to select elements from an iterable based on a condition.

👉 It keeps only those elements where the function returns True.

```python
# Syntax
# filter(function, iterable)
```

```python
nums = [1, 2, 3, 4]

result = filter(lambda x: x % 2 == 0, nums)
print(list(result))
```

```python
# Using Normal Function
def is_positive(x):
    return x > 0

nums = [-2,-1,0,3,5]
print(list(filter(is_positive, nums)))
```

```python
# Filtering Strings
words = ["Python", "", "AI", "", "ML"]
print(list(filter(lambda x: x!="", words)))
```

```python
# Filtering Dictionary
marks = {"Aman":80, "Ravi":45, "Neha":90}

passed = dict(filter(lambda x: x[1]>=50, marks.items()))
print(passed)
```

```python
# Filtering Palindromes
words = ["madam", "python", "level"]
print(list(filter(lambda x: x==x[::-1], words)))
```

```python
# Remove None Values
data = [10,None,20,None,30]
print(list(filter(lambda x: x is not None, data)))
```

```python
# Using None as Function
data = [0,1,"",None,"Python",True]
print(list(filter(None, data)))
```

```python
# filter() vs List Comprehension

# List Comprehension
nums = [1,2,3,4]
even = [x for x in nums if x%2==0]
print(even)

# filter()
even = list(filter(lambda x: x%2==0, nums))
print(even)
```

List comprehension is more readable.

```python
# filter() with Multiple Conditions
nums = [5,10,15,20,25]
result = list(filter(lambda x: x>10 and x<25, nums))
print(result)
```

```python
# filter() with Custom Objects
students = [
    {"name":"Aman","marks":80},
    {"name":"Ravi","marks":40}
]

passed = list(filter(lambda x: x["marks"]>=50, students))
print(passed)
```
