# Reduce Function
reduce() is used to reduce an iterable into a single value by applying a function cumulatively.
It repeatedly takes two elements, applies a function, and carries the result forward.

```python
from functools import reduce
reduce(lambda a, b: a + b, [1, 2, 3, 4])
```

```python
# Basic Usage with a Named Function
def add(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
reduce(add, a)
```

```python
# Using reduce() with a Lambda Function
a = [1, 2, 3, 4, 5]
reduce(lambda x, y: x * y, a)
```

```python
# Reduce with Initial Value
nums = [1, 2, 3]
reduce(lambda a, b: a + b, nums, 10)
```

```python
# Product of All Elements
nums = [1, 2, 3, 4]
reduce(lambda a, b: a * b, nums)
```

```python
# find maximum element
nums = [10, 25, 5, 30]
reduce(lambda a, b: a if a > b else b, nums)
```
