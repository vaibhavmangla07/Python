# DICTIONARY IN PYTHON
What is a Dictionary?

A dictionary is a collection of key–value pairs.

- Ordered (Python 3.7+)

- Mutable

- No duplicate keys

- Keys must be immutable

- Written using curly braces {}

```python
student = {
    "name": "Vaibhav",
    "age": 21,
    "course": "Python"
}
print(student)
print(type(student))
```

```python
print(student['name'])
print(student.get('name'))
```

```python
# empty dictionary
empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```

```python
mixed_dict = {
    "name": "Vaibhav",
    1: [1, 2, 3],
    "marks": {
        "Math": 90,
        "Science": 85
    }
}
print(mixed_dict)
print(type(mixed_dict))
```

```python
# dict() constructor
another_dict = dict(name="Vaibhav", age=21, course="Python")
print(another_dict)
print(type(another_dict))
```

```python
# Add or Update Elements
student["city"] = "Chandigarh"
student["age"] = 21
```

```python
student
```

```python
# Removing Elements
# pop()
print(student.pop("age"))

# popitem()
print(student.popitem())

# del
print("Deleted item:", student["name"])
del student["name"]

# clear()
print(student.clear())
```

```python
student
```

```python
student = {
    "name": "Vaibhav",
    "age": 21,
    "course": "Python", 
    "city": "Chandigarh"
}
print(student)
print(type(student))
```

```python
# Dictionary Methods
print(student.keys())
print(student.values())
print(student.items())
```

```python
# Loop Through Dictionary
for key, value in student.items():
    print(key, value)
```

```python
# Check Key Exists
print("name" in student)
```

```python
# Dictionary Length
print(len(student))
```

```python
# Nested Dictionary
nest_students = {
    "s1": {"name": "Aman", "age": 20},
    "s2": {"name": "Ravi", "age": 22}
}
print(nest_students)
print(type(nest_students))
print(nest_students["s1"]["name"])
```

```python
# Dictionary Comprehension 🔥
squares = {x: x*x for x in range(5)}
print(squares)
```

```python
# Copy Dictionary
new_dict = student.copy()
print(new_dict)
```

```python
# Merge Dictionaries
d1 = {"a":1}
d2 = {"b":2}

d1.update(d2)
print(d1)
```

```python
# From Keys Method
keys = ["a", "b", "c"]
d = dict.fromkeys(keys, 0)
print(d)
```

🎯 Interview Questions

Q: Why dictionary is faster?
👉 Uses hash table.

Q: Can dictionary have duplicate values?
👉 Yes, but keys must be unique.

Q: Can tuple be dictionary key?
👉 Yes.

Dict vs List
| Feature   | Dictionary    | List        |
| --------- | ------------- | ----------- |
| Access    | Key-based     | Index-based |
| Speed     | Faster lookup | Slower      |
| Structure | Key–Value     | Values only |
