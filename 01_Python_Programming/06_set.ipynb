# SET IN PYTHON
What is a Set?
A set is a collection of:

- Unordered elements

- No duplicate values

- Mutable

- Elements must be immutable

- Written using curly braces {}

```python
s = {1, 2, 3, 4}
print(s)
print(type(s))
```

```python
s = set()   # empty set
s
```

```python
# Set from List
lst = [1, 2, 2, 3]
s = set(lst)
print(s)
```

```python
# add element to set
s.add(5)
print(s)
```

```python
# update set 
s.update([6, 7, 8])
print(s)
```

### Removing Elements

```python
# remove element
s.remove(3)
print(s)
```

```python
# discard element
s.discard(5)
print(s)
```

```python
# pop element
popped_element = s.pop()
print("Popped Element:", popped_element)
print(s)
```

```python
# clear set
s.clear()
print(s)
```

### Set Operations 🔥

```python
# Union
a = {1, 2}
b = {2, 3}
print(a | b)
```

```python
# Intersection
print(a & b)
```

```python
# Difference
print(a - b)
```

```python
# Symmetric Difference
print(a ^ b)
```

-------------------------------

```python
# Subset & Superset
a = {1,2}
b = {1,2,3}

print(a.issubset(b))
print(b.issuperset(a))
```

```python
# Loop Through Set
s = {10, 20, 30, 40}
for i in s:
    print(i)
```

### Set Methods 🔥

```python
# length of set
print(len(s))
```

```python
# maximum element
print(max(s))
```

```python
# minimum element
print(min(s))
```

```python
# sum of elements
print(sum(s))
```

```python
# Frozen Set (Immutable Set)
fs = frozenset([1,2,3])
print(fs)
print(type(fs))
```

### 🔥 Set vs List

| Feature    | Set           | List      |
| ---------- | ------------- | --------- |
| Duplicates | ❌ Not allowed | ✅ Allowed |
| Order      | ❌ No          | ✅ Yes     |
| Indexing   | ❌ No          | ✅ Yes     |
| Speed      | Faster        | Slower    |
