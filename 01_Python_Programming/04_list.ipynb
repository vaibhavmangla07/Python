# List in Python
What is a List?

A list is a collection of items that is:

- Ordered
- Mutable (changeable)
- Allows duplicates
- Can store multiple data types
- Written using square brackets []

```python
lst = [1, 2, 3, 4]
print(lst)
print(type(lst))
```

```python
# mixed data types in list
lst2 = [1, "Hello", 3.4, True]
print(lst2)
```

```python
# empty list
empty_lst = []
print(empty_lst)
print(type(empty_lst))
```

```python
# using list() constructor
lst3 = list((5, 6, 7))
print(lst3)
```

### Indexing and Slicing

```python
# Indexing
lst = [10, 20, 30]
print(lst[1])
```

```python
# Negative Indexing
print(lst[-1])
```

```python
# Slicing
lst = [1, 2, 3, 4, 5]
print(lst[0:2])
print(lst[:2])
print(lst[1:])
print(lst[:])
print(lst[::2])
```

```python
# List are mutable
lst = [1, 2, 3]
lst[0] = 10
print(lst)
```

## List Methods

```python
# append()
lst = [1, 2]
lst.append(3)
lst
```

```python
# extend()
lst.extend([4, 5])
lst
```

```python
lst1 = [1, 2]
lst2 = [3, 4]
lst1.extend(lst2)
lst1
```

```python
# insert()
lst = [1, 2, 4]
lst.insert(2, 3)  # insert 3 at index 2
lst
```

```python
# remove()
lst = [1, 2, 3, 4, 3, 5]
lst.remove(3)  # remove first occurrence of 3
lst
```

```python
# pop()
lst = [1, 2, 3, 4, 5]
lst.pop()  # removes and returns last item
lst
```

```python
# clear()
lst.clear()
lst
```

------------------------------------------------------------------------------

```python
# search element
lst = [10, 20, 30, 40, 50]
print(30 in lst)
print(60 in lst)
```

```python
# list functions and methods
lst = [4, 2, 8, 1]

print(len(lst))
print(max(lst))
print(min(lst))
print(sum(lst))
print(sorted(lst))
print(lst.sort())
```

```python
# loop through list
lst = [1, 2, 3, 4, 5]
for item in lst:
    print(item)
```

```python
# List Comprehension 🔥
squares = [x**2 for x in range(1, 6)]
print(squares)
```

```python
# List Comprehension 🔥
# with condition
even = [x for x in range(10) if x % 2 == 0]
print(even)
```

```python
# Nested list
matrix = [[1,2], [3,4]]

for row in matrix:
    print(row)
```

```python
# access 
print(matrix[0][1])
```

```python
# Copying a list -- Shallow Copy
a = [1, 2, 3]
b = a.copy()
print(b)
```

```python
# Using Slicing
c = a[:]
print(c)
```

```python
# Joining Lists
a = [1,2]
b = [3,4]
c = a + b
print(c)
```

```python
# Repeating List
lst = [1,2]
print(lst * 3)
```

```python
# Reverse List
lst = [1,2,3,4,5]
lst.reverse()
print(lst)
print(lst[::-1])
```

```python
# Sorting List
lst.sort()
lst.sort(reverse=True)
print(lst)
```

```python
# Delete Elements
del lst[1]
del lst
```
