# Lambda Function
A lambda function is a small anonymous function (function without a name) used for short, one-line operations.

```python
# Simple lambda Function
square = lambda x: x * x
print(square(5))
```

```python
# lambda function with multiple arguments
add = lambda a, b: a + b
print(add(3, 7))
```

```python
# lambda function with no arguments
greet = lambda: "Hello, Lambda!"
print(greet())
```

```python
# lambda function with conditional expression
check_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(check_even(7))
```

```python
# lambda with map() : Used to apply a function to every element of an iterable like list, tuple etc.
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(squared)
```

```python
# lambda with filter() : Used to filter elements from an iterable based on a condition.
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

```python
# lambda with reduce() : Used to apply a rolling computation to sequential pairs of values in an iterable.
from functools import reduce
product = reduce(lambda x, y: x * y, numbers)
print(product)
```

```python
# lambda with sorted() : Used to sort an iterable based on a key function.
points = [(1, 2), (4, 1), (5, -1), (2, 3)]
sorted_points = sorted(points, key=lambda point: point[1])
print(sorted_points)
```

```python
# lamda with list comprehension : Used to create a new list by applying an expression to each item in an iterable.
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    print(i())
```
