# PYTHON Basics

 1. Printing function

```python
print("Hello, World!")
print(3)
print(5 + 7)
print("5" + "7")
print("5" * 3)
# print("5" + 7) --> This will raise a TypeError
```

 2. Data Types and Variables

```python
a = 10
b = 3.5
c = "Python"
d = True
e = None
f = [1, 2, 3, 4, 5]
g = (1, 2, 3)
h = { "name": "Alice", "age": 25 }
i = {1, 2, 3, 4, 5}

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))
print(type(i))
```

 3. Type casting & Type conversion

```python
# Type Casting
x = "100"
y = int(x)

print(y, type(y))

# Type Conversion
p = 5
q = float(p)
print(q, type(q))
 
```

 4. Input from user

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print(f"My name is {name} and I am {age} years old")
```

5. Arithmetic Operations

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
```

6. Relational Operators

```python
a = 13
b = 33

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

7. Logical Operators

```python
a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

8. Bitwise Operators

```python
a = 10
b = 4

print(a & b)
print(a | b)
print(~a)
print(a ^ b)
print(a >> 2)
print(a << 2)
```

9. Conditional Statements

```python
num = 12

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

10. Control Statements

```python
# for loop
for i in range(5):
    print(i)
```

```python
# while loop
i = 1
while i <= 5:
    print(i)
    i += 1
```

11. Break and Continue

```python
# Break and Continue
print("Break Example:")
for i in range(10):
    if i == 5:
        break
    print(i)    

print("\nContinue Example:")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i) 
```

# Functions

```python
# Function with no arguments and no return value
def fun():
    print("Hello World!")
fun()
```

```python
# Function with arguments and no return value
def add(a, b):
    print("Sum:", a + b)

add(10, 20)
```

```python
# Function with no arguments but with return value
def fun():
    return 3.14

value = fun()
print(value)
```

```python
# Function with arguments and return value
def multiply(a, b):
    return a * b
multiply(5, 4)
```

```python
# Local and Global Variables

num1 = 10  # Global variable
def my_function():
    num2 = 5  # Local variable
    print("Inside function, num2 =", num2)
    print("Inside function, num1 =", num1)
my_function()
print("Outside function, num1 =", num1)
# print("Outside function, num2 =", num2)  # This will raise an error
```

# Recursion

```python
# Recursion - Factorial of a number
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print(factorial(5))
```

```python
# Recursion - Fibonacci series
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
```
