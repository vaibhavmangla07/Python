# Tuple in Python
### What is a Tuple?

A tuple is a collection of elements that is:

- Ordered
- Immutable (cannot be changed)
- Allows duplicate values
- Written using parentheses ()

Tuples are immutable; their elements cannot be changed after creation.

Hence, methods that modify the tuple (like append, remove, etc.) are not available.

Tuples support indexing and slicing similar to lists.

```python
t = (1, 2, 3, 4)
print(t)
print(type(t))
```

```python
# mixed data types in tuple
t2 = (1, "Hello", 3.4)
t2
```

```python
t = (5)
print(type(t))
```

```python
t = (5,)
print(type(t))
```

```python
t = 1, 2, 3
print(t)
print(type(t))
```

### Indexing & Slicing

```python
# Indexing
t = (10, 20, 30)
t[1]
```

```python
# Negative Indexing
t[-1]
```

```python
t = (1, 2, 3, 4, 5)
t[1:4]
```

```python
t = (1, 2, 3)
# t[0] = 10 --> Error
```

```python
# Tuple methods
print(t.count(2))
print(t.index(3))
```

```python
# Tuple packing and unpacking

t = (1, 2, 3)
a, b, c = t
a, b, c
```

```python
# Nested tuples
t = (1, 2, (3, 4), 5)
t[2][1]
```

```python
# Convert List to Tuple
l = [1, 2, 3]
t = tuple(l)
print(t)
print(type(t))
```

```python
# Loop through a tuple
t = (1, 2, 3)
for i in t:
    print(i)
```

Why Use Tuple Instead of List?

- Faster
- Data safety (immutable)
- Used as dictionary keys
- Memory efficient
