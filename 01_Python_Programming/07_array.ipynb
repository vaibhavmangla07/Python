# ARRAY IN PYTHON 
✅ What is an Array?

An array is a collection of elements of the same data type stored in continuous memory locations.

📌 Python does not have built-in arrays like C/C++, but we use:
1. List (commonly used as array)
2. array module
3. NumPy arrays (advanced)

```python
arr = [10, 20, 30, 40]
print(arr)
```

```python
# Access Elements
print(arr[0])
print(arr[-1])
```

```python
# Array Using array Module

import array

arr = array.array('i', [1, 2, 3, 4])
print(arr)
```

| Code | Type    |
| ---- | ------- |
| 'i'  | Integer |
| 'f'  | Float   |
| 'd'  | Double  |
| 'u'  | Unicode |


### Array Operations🔥

```python
# Append
arr.append(5)
print(arr)
```

```python
# insert
arr.insert(2, 15)
print(arr)
```

```python
# remove 
arr.remove(3)
print(arr)
```

```python
# pop
arr.pop()
print(arr)
```

```python
# Loop Through Array
for i in arr:
    print(i)
```

```python
# Reverse Array
arr.reverse()
print(arr)
```

```python
# Find Index
print(arr.index(2))
```

```python
# Update Value
arr[0] = 100
print(arr)
```

```python
# Length of Array
print(len(arr))
```

### Array vs List
| Feature   | Array    | List     |
| --------- | -------- | -------- |
| Data Type | Same     | Mixed    |
| Speed     | Faster   | Slower   |
| Memory    | Less     | More     |
| Module    | Required | Built-in |


```python
# NumPy Arrays (Advanced)
import numpy as np

arr = np.array([1,2,3,4])
print(arr)
```
